```python
# Quick FileMaker access smoke test (read-only)
from pathlib import Path
import sys

# Ensure the repo root is importable when running from notebooks/
repo_root = Path.cwd().resolve().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from bs_fmp_sdk import FileMakerClient, Layouts, load_config

config = load_config(repo_root / ".env")
print(f"Host: {config.host}")
print(f"Database: {config.database}")

client = FileMakerClient(config)
try:
    if not client.ping(Layouts.PROJECTS):
        raise RuntimeError("Ping failed. Check credentials, host, and layout access.")

    records = client.get_records(layout_name=Layouts.PROJECTS, limit=1)
    print("Connected to FileMaker successfully.")
    print(f"Fetched {len(records)} record(s) from {Layouts.PROJECTS}.")

    if records:
        sample_keys = sorted(records[0].keys())[:10]
        print("Sample field keys:", sample_keys)
finally:
    client.logout()
    print("Session closed.")
```
